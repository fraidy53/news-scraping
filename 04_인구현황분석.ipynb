1. 지역별 총인구수 Top5

```python
import pandas as pd

# 1. 데이터 불러오기
df = pd.read_csv('data/인구현황.csv')
df = df[df['행정기관'] != '전국']

# 2. 총인구수 기준 내림차순 정렬
df_sorted = df.sort_values(by='총인구수', ascending=False)

# 3. 상위 5개 추출
top5 = df_sorted.head(5)

# 4. 필요한 컬럼만 출력
result = top5[['행정기관','총인구수']]

result
```

1-1. 지역별 총인구수 plot

```python
def show_pop_area():
    df = pd.read_csv('data/인구현황.csv')
    df = df[df['행정기관'] != '전국']
    figure, axes1 = plt.subplots(nrows=1, ncols=1) # 2행 1열
    figure.set_size_inches(18, 12)
    
    pop_plot = sns.barplot(data=df.sort_values(by='총인구수', ascending=False), x='행정기관', y='총인구수', ax=axes1, hue='행정기관')
    axes1.set_xlabel('')
    pop_plot.set_title(f'지역별 총인구수')

    plt.xticks(rotation=45)
    plt.tight_layout()
show_pop_area()
```

3. 남녀 비율 분석

```python
# 1. 데이터 불러오기
df = pd.read_csv('data/인구현황.csv')
df = df[df['행정기관'] != '전국']

# 2. 남녀 비율 기준 내림차순 정렬
df_sorted = df.sort_values(by='남여 비율', ascending=False).reset_index()

# 3.  추출
top5 = df_sorted

# 4. 필요한 컬럼만 출력
result = top5[['행정기관','남여 비율']]

result
```

```python
def show_pop_area2():
    df = pd.read_csv('data/인구현황.csv')
    df = df[df['행정기관'] != '전국']
    figure, axes1 = plt.subplots(nrows=1, ncols=1) # 2행 1열
    figure.set_size_inches(18, 12)
    
    pop_plot = sns.barplot(data=df.sort_values(by='남여 비율', ascending=False), x='행정기관', y='남여 비율', ax=axes1, hue='행정기관')
    pop_plot.set_title(f'지역별 남녀 비율')
    axes1.set_xlabel('')
    axes1.axhline(y=1.0, color='red', linestyle='--', label='남녀 1:1 비율')
    axes1.legend()
    plt.xticks(rotation=45)
    plt.tight_layout()
show_pop_area2()
```

5-1.    남초 vs 여초 개수 Plot  ( seaborn의 countplot )
        지역별 세대수 Plot ( seaborn의 barplot )

```python
def show_pop_area3():
    df = pd.read_csv('data/인구현황.csv')
    df['남초여초'] = df['남여 비율'].apply(lambda x: '남초' if x > 1.0 else('여초' if x <1.0 else '동일'))
    df = df[df['행정기관'] != '전국']
    figure, (axes1, axes2) = plt.subplots(nrows=2, ncols=1) # 2행 1열
    figure.set_size_inches(18, 12)
    
    pop_plot = sns.countplot(data=df.sort_values(by='남초여초'), x='남초여초', ax=axes1, hue='남초여초')
    pop_plot.set_title(f'남초 vs 여초 지역 개수')

    area_plot = sns.barplot(data=df.sort_values(by='세대수', ascending=False), x='행정기관', y='세대수', ax=axes2, hue='행정기관')
    area_plot.set_title(f'지역별 세대수')
    
    plt.tight_layout()
show_pop_area3()
```

6. 남자 인구 vs 여자 인구 비교 Plot(선택)

```python
def show_male_female_compare():
    df = pd.read_csv('data/인구현황.csv')
    df = df[df['행정기관'] != '전국']

    # 정렬 (총 인구 기준으로 하면 보기 좋음)
    

    figure, ax = plt.subplots(figsize=(18, 10))

    # 데이터 구조를 long 형태로 변환 (그래야 hue 사용 가능)
    df_melted = df.melt(id_vars='행정기관', value_vars=['남자 인구수', '여자 인구수'], var_name='성별', value_name='인구수')

    sns.barplot(data=df_melted, x='행정기관', y='인구수', hue='성별',ax=ax)

    ax.set_title('지역별 남자 인구 vs 여자 인구 비교')
    ax.set_ylabel('')
    ax.tick_params(axis='x', rotation=45)

    plt.tight_layout()
show_male_female_compare()
```
